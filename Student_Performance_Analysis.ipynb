```python
import pandas as pd

data = {
    "Student": ["Amina", "Rahim", "Karim", "Nadia", "Sadia","Hasan", "Fatima", "Imran", "Rafi", "Maya"],
    "Math": [85, 72, 90, 88, 65, 78, 95, 60, 82, 91],
    "English": [78, 80, 85, 92, 70, 75, 89, 68, 84, 94],
    "Science": [90, 75, 88, 85, 72, 80, 96, 65, 79, 93]
}

df = pd.DataFrame(data)

df["Avarage"] = df[["Math","English", "Science"]].mean(axis=1)
print(df)
```

```python
#View the first five students
display(df.head())

#Check the size of the dataset
print("Number of Students: ", df.shape[0])
print("Number of Columns: ", df.shape[1])

#Check for missing values
print("\nMissing Values: ")
print(df.isnull().sum())

#Summary Statistics
print("\nSummary Statistics: ")
display(df.describe().round(2))
```

```python
# Average score for each subject
subjects = ["Math", "English", "Science"]

print("Average score by subject:")
print(df[subjects].mean().round(2))

# Find the highest-performing student
top_student = df.loc[df["Average"].idxmax()]

print("\nTop-performing student:")
print("Name:", top_student["Student"])
print("Average:", round(top_student["Average"], 2))

# Find students with an average of 80 or above
high_performers = df[df["Average"] >= 80]

print("\nStudents averaging 80 or above:")
display(high_performers[["Student", "Average"]].round(2))
```

```python
import matplotlib.pyplot as plt

subject_avarages = df[subjects].mean()

plt.figure(figsize=(8,5))
subject_avarages.plot(kind="bar")

plt.title("Average Score by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Score")
plt.ylim(0, 100)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
